#### Machine Learning vs Reinforce Learning

머신 러닝
* 통계적 알고리즘(Statical algorithms)을 통해 컴퓨터가 스스로 여러 데이터를 학습하여, 새로운 데이터에 대해서도 일반화를 할 수 있도록 하는 것
* 어떤 데이터 x가 있을 때, 결과 y와의 관계 f(x)=y를 확률적으로 학습함
* Linear regression, Decision tree, Random forest, K-nearest neighbor(KNN), Support vector machine(SVM), Multilayer perceptron(MLP) 등을 포함함

| 항목         | 기계학습 (Machine Learning)          | 강화학습 (Reinforcement Learning)      |
|--------------|--------------------------------------|----------------------------------------|
| 학습 방법     | 입력 + 정답 데이터를 보고 학습         | 보상을 기반으로 trial & error 방식 학습 |
| 피드백       | 정답이 존재함 (예: 고양이 vs 개)        | 정답 없음, 보상/벌점만 존재            |
| 목표         | 예측 정확도 높이기                    | **누적 보상**을 최대화                 |
| 학습 환경     | 정적인 데이터셋                      | **환경과의 상호작용**이 필요함         |
| 주 사용 분야  | 이미지 분류, 스팸 필터링, 예측 모델     | 게임 AI, 로봇 제어, 자율 주행          |

##### 기계 학습에서 데이터셋의 표현

* 샘플은 특징 벡터(feature vector)와 레이블(label)로 표현한다.
* 특징 벡터는 x로 표기(d = 특징의 개수 = 벡터의 차원)
* 레이블은 0,1,2,…,c-1의 값 또는 1,2,…,c-1,c의 값 또는 원핫 코드(한 요소만 1인 이진열 e.g. (1,0,0))
* 용어
  * instance = 샘픔
  * feature = 특징
  * class = 부류

```python
from sklearn import datasets

d = datasets.load_iris()
# print(d.DESCR)
print(d.data, d.target)
```

```python
for i in range(0, len(d.data)):
  print(i+1, d.data[i], d.target[i]) # target is the class label
```

SVM(Support Vector Machine)
* 데이터를 분리하는 가장 좋은 선(혹은 초평면,hyperplane)을 찾는 알고리즘이다.
* 구체적으로, 각 클래스에서 가장 가까운 데이터 포인트들(=서포트 벡터)과의 마진(margin)을 최대화하는 선을 의미한다.

> Support Vectors : 초평면에 가장 가까이 있는 데이터 포인트

Classifier (분류기) 주요 기능
1. 데이터 분류: 입력된 데이터를 미리 정의된 클래스 중 하나로 분류. 
2. 패턴 인식: 데이터 내의 패턴을 학습하여 분류를 수행. 
3. 예측: 새로운 데이터가 입력되었을 때 어떤 클래스에 속할지 예측. 
4. 학습: 분류 모델은 학습을 통해 정확도를 향상시킨다.

Hyper Parameter
모델의 동작을 제어하는 데 쓰는 변수로 모델 학습을 시작하기 전에 설정해야 좋은 성능을 얻을 수 있다. 최적의 하이퍼 매개변수 값을 자동으로 설정하는 일을 하이퍼 매개변수 최적화(hyper parameter optimization)라 한다.

train set과 test set
* Train Set : 기계 학습 모델을 학습하는데 쓰는 데이터로 Feature vector와 label 정보를 모두 제공
* Test Set : 학습을 마친 모델의 성능을 측정하는데 쓰는 데이터로서 예측할 때는 feature vector 정보만 제공하고, 예측 결과를 가지고 정확률을 측정할 때 label 정보를 사용

| 구분                  | 특징 추출 (Feature Extraction)           | 분류기 학습 (Classifier)           |
|-----------------------|-------------------------------------------|------------------------------------|
| **전통적 머신러닝**    | 사람이 직접 만들거나, 통계적으로 설계        | SVM, RandomForest 등으로 따로 학습  |
| **딥러닝**             | **네트워크가 자동으로 학습**함               | 네트워크의 마지막 계층에서 같이 학습 |


```python
from sklearn import svm

# 분류기 객체 생성
s = svm.SVC(gamma=0.1, C=10) # Support Vector Classifier, C는 하이퍼 파라미터
s.fit(d.data,d.target) # 훈련 집합

new_d=[[6.4, 3.2, 6.0, 2.5], [7.1, 3.1, 4.7, 1.35]] # 새로운 샘플 데이터 설명

res = s.predict(new_d)
print("새로운 2개의 샘플의 class", res)
```
